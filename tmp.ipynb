```python
from datasets import load_dataset
from datasets import DatasetDict

# load HuggingFaceM4/A-OKVQA@train and make a new dataset with only 10 rows
dataset = load_dataset("HuggingFaceM4/A-OKVQA")
dataset_train = dataset["train"].select(range(3000))
dataset_debug = dataset["train"].select(range(32))
dataset = DatasetDict({"train": dataset_train, "debug": dataset_debug})
print(dataset)
# push 
dataset.push_to_hub("Amasia-MLLM/A-OKVQA")
```

```python
import torch
from tensordict import TensorDict

a = TensorDict({"a": torch.rand(2, 10), "b": torch.rand(2, 10)}, batch_size=2)
b = TensorDict({"a": torch.rand(2, 10), "b": torch.rand(2, 10)}, batch_size=2)
```

```python
# Concatenate TensorDicts directly using the cat method
c = TensorDict.cat([a, b], dim=0)
```
